```python
import os
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = "/content/drive/MyDrive/llm2"
RESULTS_DIR = f"{BASE_DIR}/results"
LOSS_DIR = f"{RESULTS_DIR}/loss_curves"

os.makedirs(LOSS_DIR, exist_ok=True)
```

```python
loss_data = {
    "SmolLM2-360M": {
        "epoch": [1, 2, 3],
        "train_loss": [2.284184, 4.783071, 7.057727],
        "val_loss": [4.882849, 4.767120, 4.713677]
    },
    "TinyLlama-1.1B": {
        "epoch": [1, 2, 3],
        "train_loss": [2.049165, 2.130613, 2.055511],
        "val_loss": [2.385674, 2.178175, 2.075848]
    },
    "Qwen2.5-1.5B": {
        "epoch": [1, 2, 3],
        "train_loss": [2.601168, 2.851104, 2.858479],
        "val_loss": [3.234486, 2.997751, 2.916851]
    }
}
```

```python
for model_name, values in loss_data.items():
    df_loss = pd.DataFrame(values)

    plt.figure(figsize=(7, 5))
    plt.plot(df_loss["epoch"], df_loss["train_loss"], marker="o", label="Training Loss")
    plt.plot(df_loss["epoch"], df_loss["val_loss"], marker="o", label="Validation Loss")

    plt.title(f"{model_name} Loss Curve")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.xticks(df_loss["epoch"])
    plt.legend()
    plt.grid(True)

    safe_name = model_name.replace(".", "").replace("-", "_")
    save_path = f"{LOSS_DIR}/{safe_name}_loss_curve.png"

    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print("Kaydedildi:", save_path)
```

```python
plt.figure(figsize=(8, 5))

for model_name, values in loss_data.items():
    df_loss = pd.DataFrame(values)
    plt.plot(df_loss["epoch"], df_loss["val_loss"], marker="o", label=model_name)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.xticks([1, 2, 3])
plt.legend()
plt.grid(True)

save_path = f"{LOSS_DIR}/validation_loss_comparison.png"
plt.savefig(save_path, dpi=300, bbox_inches="tight")
plt.show()

print("Kaydedildi:", save_path)
```

```python
rows = []

for model_name, values in loss_data.items():
    for i in range(len(values["epoch"])):
        rows.append({
            "model": model_name,
            "epoch": values["epoch"][i],
            "train_loss": values["train_loss"][i],
            "val_loss": values["val_loss"][i]
        })

loss_df = pd.DataFrame(rows)

loss_csv_path = f"{LOSS_DIR}/loss_values.csv"
loss_df.to_csv(loss_csv_path, index=False, encoding="utf-8-sig")

display(loss_df)
print("Kaydedildi:", loss_csv_path)
```
