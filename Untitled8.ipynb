```python
import os
import pandas as pd
import numpy as np

BASE_DIR = "/content/drive/MyDrive/llm2"
DATASET_DIR = f"{BASE_DIR}/datasets"
SEED_DIR = f"{BASE_DIR}/seed_runs"
RESULTS_DIR = f"{BASE_DIR}/results"

os.makedirs(RESULTS_DIR, exist_ok=True)

print("DATASET_DIR:", os.path.exists(DATASET_DIR))
print("SEED_DIR:", os.path.exists(SEED_DIR))
print("RESULTS_DIR:", os.path.exists(RESULTS_DIR))
```

```python
files = {
    "SmolLM2-360M": f"{SEED_DIR}/SmolLM2-360M_3seed_results.csv",
    "TinyLlama-1.1B": f"{SEED_DIR}/TinyLlama-1.1B_3seed_results.csv",
    "Qwen2.5-1.5B": f"{SEED_DIR}/Qwen2.5-1.5B_3seed_results.csv"
}

all_seed_results = []

for model_name, path in files.items():
    df_model = pd.read_csv(path)
    all_seed_results.append(df_model)

seed_df = pd.concat(all_seed_results, ignore_index=True)

display(seed_df)
```

```python
metrics = [
    "train_loss",
    "val_loss",
    "train_time_sec",
    "gpu_memory_gb",
    "rouge1",
    "rouge2",
    "rougeL",
    "bleu",
    "inference_ms"
]

summary_rows = []

for model_name, group in seed_df.groupby("model"):
    row = {"Model": model_name}

    for metric in metrics:
        mean = group[metric].mean()
        std = group[metric].std()

        row[f"{metric}_mean"] = mean
        row[f"{metric}_std"] = std
        row[f"{metric}"] = f"{mean:.4f} ± {std:.4f}"

    summary_rows.append(row)

benchmark_df = pd.DataFrame(summary_rows)

display(benchmark_df)
```

```python
model_sizes = pd.read_csv(f"{DATASET_DIR}/model_sizes.csv")

display(model_sizes)
```

```python
benchmark_df = benchmark_df.merge(
    model_sizes.rename(columns={"model": "Model"}),
    on="Model",
    how="left"
)

display(benchmark_df)
```

```python
param_map = {
    "SmolLM2-360M": "360M",
    "TinyLlama-1.1B": "1.1B",
    "Qwen2.5-1.5B": "1.5B"
}

final_table = pd.DataFrame({
    "Model": benchmark_df["Model"],
    "Parametre": benchmark_df["Model"].map(param_map),
    "FT Yöntemi": "QLoRA",
    "ROUGE-1": benchmark_df["rouge1"],
    "ROUGE-L": benchmark_df["rougeL"],
    "BLEU": benchmark_df["bleu"],
    "Eğitim Süresi (sn)": benchmark_df["train_time_sec"],
    "Inference (ms/örnek)": benchmark_df["inference_ms"],
    "GPU Bellek (GB)": benchmark_df["gpu_memory_gb"],
    "Model Boyutu (MB)": benchmark_df["model_size_MB"].round(2)
})

display(final_table)

final_table.to_csv(
    f"{RESULTS_DIR}/final_benchmark_table.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Kaydedildi:", f"{RESULTS_DIR}/final_benchmark_table.csv")
```

```python
zero_df = pd.read_csv(f"{DATASET_DIR}/zero_shot_results.csv")

display(zero_df)
```

```python
ft_mean = seed_df.groupby("model")[["rouge1", "rouge2", "rougeL", "bleu", "inference_ms"]].mean().reset_index()

zero_df["Durum"] = "Zero-shot"
ft_mean["Durum"] = "Fine-tuned"

zero_vs_ft = pd.concat([
    zero_df.rename(columns={"model": "Model"}),
    ft_mean.rename(columns={"model": "Model"})
], ignore_index=True)

zero_vs_ft = zero_vs_ft[[
    "Model", "Durum", "rouge1", "rouge2", "rougeL", "bleu", "inference_ms"
]]

display(zero_vs_ft)

zero_vs_ft.to_csv(
    f"{RESULTS_DIR}/zero_vs_finetuned_table.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Kaydedildi:", f"{RESULTS_DIR}/zero_vs_finetuned_table.csv")
```
