```python
import os
import torch
import pandas as pd

model_dirs = {
    "SmolLM2-360M": "/content/drive/MyDrive/llm2/models/smollm2_danistay_40",
    "TinyLlama-1.1B": "/content/drive/MyDrive/llm2/models/tinyllama_danistay_40",
    "Qwen2.5-1.5B": "/content/drive/MyDrive/llm2/models/qwen25_15b_danistay_40"
}

rows = []

for name, path in model_dirs.items():
    total_size = 0
    for root, dirs, files in os.walk(path):
        for f in files:
            fp = os.path.join(root, f)
            total_size += os.path.getsize(fp)

    rows.append({
        "model": name,
        "model_size_MB": total_size / (1024 * 1024)
    })

size_df = pd.DataFrame(rows)
display(size_df)

size_df.to_csv(
    "/content/drive/MyDrive/llm2/datasets/model_sizes.csv",
    index=False,
    encoding="utf-8-sig"
)
```
