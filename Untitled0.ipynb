```python
import os
import re
import pandas as pd
from sklearn.model_selection import train_test_split

KARAR_DIR = "/content/drive/MyDrive/llm2/danıstay_karar"
OZET_DIR = "/content/drive/MyDrive/llm2/danıstay_ozet"
DATASET_DIR = "/content/drive/MyDrive/llm2/datasets"

os.makedirs(DATASET_DIR, exist_ok=True)

print("Karar klasörü:", os.path.exists(KARAR_DIR))
print("Özet klasörü:", os.path.exists(OZET_DIR))
print("Dataset klasörü:", os.path.exists(DATASET_DIR))
```

```python
karar_files = sorted([f for f in os.listdir(KARAR_DIR) if f.endswith(".txt")])
ozet_files = sorted([f for f in os.listdir(OZET_DIR) if f.endswith(".txt")])

print("Karar sayısı:", len(karar_files))
print("Özet sayısı:", len(ozet_files))

print("\nKarar örnekleri:")
print(karar_files[:10])

print("\nÖzet örnekleri:")
print(ozet_files[:10])
```

```python
def normalize_name(name):
    name = name.replace(".txt", "")
    name = name.replace("_ozet_", "_")
    name = name.replace("saglik", "saglık")
    return name

def read_txt(path):
    with open(path, "r", encoding="utf-8", errors="ignore") as f:
        return f.read().strip()

ozet_map = {}

for f in ozet_files:
    key = normalize_name(f)
    ozet_map[key] = f

rows = []

for karar_file in karar_files:
    key = normalize_name(karar_file)

    if key not in ozet_map:
        print("Eşleşmeyen karar:", karar_file)
        continue

    category = key.rsplit("_", 1)[0]

    rows.append({
        "id": key,
        "category": category,
        "karar_file": karar_file,
        "ozet_file": ozet_map[key],
        "input_text": read_txt(os.path.join(KARAR_DIR, karar_file)),
        "summary": read_txt(os.path.join(OZET_DIR, ozet_map[key]))
    })

df = pd.DataFrame(rows)

print("Eşleşen örnek sayısı:", len(df))
display(df[["id", "category", "karar_file", "ozet_file"]])
```

```python
def normalize_for_search(text):
    text = text.replace("İ", "I").replace("ı", "i")
    text = text.upper()
    return text

def find_section(text, start_keywords, end_keywords=None, max_chars=2500):
    norm_text = normalize_for_search(text)

    start_idx = -1
    for keyword in start_keywords:
        idx = norm_text.find(normalize_for_search(keyword))
        if idx != -1:
            start_idx = idx
            break

    if start_idx == -1:
        return ""

    end_idx = len(text)

    if end_keywords:
        for end_keyword in end_keywords:
            idx = norm_text.find(normalize_for_search(end_keyword), start_idx + 20)
            if idx != -1:
                end_idx = min(end_idx, idx)

    section = text[start_idx:end_idx]
    section = re.sub(r"\n{3,}", "\n\n", section)
    section = re.sub(r"[ \t]+", " ", section)

    return section[:max_chars].strip()


def build_model_input(text):
    dava = find_section(
        text,
        ["Dava konusu istem", "Dava Konusu İstem", "DAVA KONUSU İSTEM"],
        ["İlk Derece Mahkemesi", "Daire kararının özeti", "Bölge İdare Mahkemesi", "TEMYİZ EDEN"],
        max_chars=1200
    )

    temyiz = find_section(
        text,
        ["TEMYİZ EDENİN İDDİALARI", "TEMYİZ EDENLERİN İDDİALARI", "DAVACININ İDDİALARI"],
        ["KARŞI TARAF", "DANIŞTAY TETKİK", "TÜRK MİLLETİ ADINA", "İNCELEME VE GEREKÇE"],
        max_chars=1200
    )

    savunma = find_section(
        text,
        ["KARŞI TARAFIN SAVUNMASI", "KARŞI TARAFIN CEVABI"],
        ["DANIŞTAY TETKİK", "TÜRK MİLLETİ ADINA", "İNCELEME VE GEREKÇE"],
        max_chars=800
    )

    maddi_olay = find_section(
        text,
        ["MADDİ OLAY", "MADDİ OLAY :"],
        ["İLGİLİ MEVZUAT", "HUKUKİ DEĞERLENDİRME", "KARAR SONUCU"],
        max_chars=1500
    )

    hukuki = find_section(
        text,
        ["HUKUKİ DEĞERLENDİRME"],
        ["KARAR SONUCU", "SONUÇ", "KARŞI OY"],
        max_chars=3000
    )

    if hukuki == "":
        hukuki = find_section(
            text,
            ["İNCELEME VE GEREKÇE"],
            ["KARAR SONUCU", "SONUÇ", "KARŞI OY"],
            max_chars=3000
        )

    sonuc = find_section(
        text,
        ["KARAR SONUCU", "SONUÇ"],
        ["KARŞI OY"],
        max_chars=800
    )

    combined = f"""
DAVA KONUSU:
{dava}

TEMYİZ İDDİALARI:
{temyiz}

KARŞI TARAF SAVUNMASI:
{savunma}

MADDİ OLAY:
{maddi_olay}

HUKUKİ DEĞERLENDİRME:
{hukuki}

KARAR SONUCU:
{sonuc}
"""

    combined = re.sub(r"\n{3,}", "\n\n", combined)
    combined = re.sub(r"[ \t]+", " ", combined)

    return combined.strip()
```

```python
df["model_input"] = df["input_text"].apply(build_model_input)

df["input_len"] = df["input_text"].apply(len)
df["model_input_len"] = df["model_input"].apply(len)
df["summary_len"] = df["summary"].apply(len)

display(df[["id", "category", "input_len", "model_input_len", "summary_len"]])
display(df[["model_input_len", "summary_len"]].describe())
```

```python
def fallback_input(text, max_chars=3500):
    text = re.sub(r"\n{3,}", "\n\n", text)
    text = re.sub(r"[ \t]+", " ", text)
    return text[:max_chars].strip()

short_mask = df["model_input_len"] < 500

print("Kısa yakalanan örnek sayısı:", short_mask.sum())
display(df.loc[short_mask, ["id", "category", "model_input_len"]])

df.loc[short_mask, "model_input"] = df.loc[short_mask, "input_text"].apply(fallback_input)
df["model_input_len"] = df["model_input"].apply(len)

display(df[["id", "category", "input_len", "model_input_len", "summary_len"]])
```

```python
df_final = df[["id", "category", "karar_file", "ozet_file", "model_input", "summary"]].copy()

final_path = f"{DATASET_DIR}/danistay_40_teacher_final.csv"

df_final.to_csv(final_path, index=False, encoding="utf-8-sig")

print("Kaydedildi:", final_path)
print("Toplam örnek:", len(df_final))
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    df_final,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

train_df.to_csv(f"{DATASET_DIR}/train_40.csv", index=False, encoding="utf-8-sig")
val_df.to_csv(f"{DATASET_DIR}/validation_40.csv", index=False, encoding="utf-8-sig")
test_df.to_csv(f"{DATASET_DIR}/test_40.csv", index=False, encoding="utf-8-sig")

print("Bölünmüş dosyalar kaydedildi.")
```

```python
import pandas as pd
import torch

DATASET_DIR = "/content/drive/MyDrive/llm2/datasets"

train_df = pd.read_csv(f"{DATASET_DIR}/train_40.csv")
val_df = pd.read_csv(f"{DATASET_DIR}/validation_40.csv")
test_df = pd.read_csv(f"{DATASET_DIR}/test_40.csv")

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
def make_text(row):
    return f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}

Özet:
{row["summary"]}"""

train_texts = train_df.apply(make_text, axis=1).tolist()
val_texts = val_df.apply(make_text, axis=1).tolist()
test_texts = test_df.apply(make_text, axis=1).tolist()

print("Train text:", len(train_texts))
print(train_texts[0][:1000])
```

```python
from torch.utils.data import Dataset

class TextDataset(Dataset):
    def __init__(self, texts, tokenizer, max_length=2048):
        self.texts = texts
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        enc = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_length,
            padding="max_length",
            return_tensors="pt"
        )

        input_ids = enc["input_ids"].squeeze()
        attention_mask = enc["attention_mask"].squeeze()

        return {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            "labels": input_ids.clone()
        }
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "HuggingFaceTB/SmolLM2-360M-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    torch_dtype=torch.float16
)
```

```python
train_dataset = TextDataset(train_texts, tokenizer, max_length=2048)
val_dataset = TextDataset(val_texts, tokenizer, max_length=2048)

print(len(train_dataset))
print(train_dataset[0]["input_ids"].shape)
```

```python
from peft import LoraConfig, get_peft_model

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
```

```python
from transformers import TrainingArguments, Trainer

OUTPUT_DIR = "/content/drive/MyDrive/llm2/models/smollm2_danistay_40"

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    num_train_epochs=3,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    logging_steps=1,
    eval_strategy="epoch",
    save_strategy="epoch",
    fp16=True,
    report_to="none",
    seed=42
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset
)

trainer.train()
```

```python
model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)

print("Model kaydedildi:", OUTPUT_DIR)
```

```python
def make_prompt_only(row):
    return f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}

Özet:
"""

def generate_summary(text, max_new_tokens=250):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=2048
    ).to(model.device)

    model.eval()

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.2,
            no_repeat_ngram_size=4,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = outputs[0][inputs["input_ids"].shape[-1]:]
    return tokenizer.decode(generated, skip_special_tokens=True).strip()
```

```python
predictions = []
references = []

for i, row in test_df.iterrows():
    prompt = make_prompt_only(row)
    pred = generate_summary(prompt)

    predictions.append(pred)
    references.append(row["summary"])

    print("="*80)
    print("ID:", row["id"])
    print("\nMODEL ÖZETİ:\n", pred)
    print("\nREFERANS ÖZET:\n", row["summary"])
```

```python
times = []

for i, row in test_df.iterrows():
    prompt = make_prompt_only(row)

    start = time.time()
    _ = generate_summary(prompt)
    end = time.time()

    times.append((end - start) * 1000)

avg_inference_ms = sum(times) / len(times)

print("Ortalama inference süresi ms/örnek:", avg_inference_ms)
```
