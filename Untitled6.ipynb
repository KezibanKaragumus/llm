```python
import os, gc, time, random
import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset
from transformers import set_seed

DATASET_DIR = "/content/drive/MyDrive/llm2/datasets"

train_df = pd.read_csv(f"{DATASET_DIR}/train_40.csv")
val_df = pd.read_csv(f"{DATASET_DIR}/validation_40.csv")
test_df = pd.read_csv(f"{DATASET_DIR}/test_40.csv")

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

def make_text(row):
    return f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}

Özet:
{row["summary"]}"""

def make_prompt_only(row):
    return f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}

Özet:
"""

train_texts = train_df.apply(make_text, axis=1).tolist()
val_texts = val_df.apply(make_text, axis=1).tolist()

class TextDataset(Dataset):
    def __init__(self, texts, tokenizer, max_length=1024):
        self.texts = texts
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        enc = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_length,
            padding="max_length",
            return_tensors="pt"
        )

        input_ids = enc["input_ids"].squeeze()

        return {
            "input_ids": input_ids,
            "attention_mask": enc["attention_mask"].squeeze(),
            "labels": input_ids.clone()
        }
```

```python
import evaluate

rouge = evaluate.load("rouge")
bleu = evaluate.load("sacrebleu")

def generate_summary(model, tokenizer, text, max_len=1024, max_new_tokens=250):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=max_len
    ).to(model.device)

    model.eval()

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.2,
            no_repeat_ngram_size=4,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = outputs[0][inputs["input_ids"].shape[-1]:]
    return tokenizer.decode(generated, skip_special_tokens=True).strip()

def evaluate_model(model, tokenizer):
    predictions = []
    references = []
    times = []

    for _, row in test_df.iterrows():
        prompt = make_prompt_only(row)

        start = time.time()
        pred = generate_summary(model, tokenizer, prompt)
        end = time.time()

        predictions.append(pred)
        references.append(row["summary"])
        times.append((end - start) * 1000)

    rouge_result = rouge.compute(
        predictions=predictions,
        references=references
    )

    bleu_result = bleu.compute(
        predictions=predictions,
        references=[[r] for r in references]
    )

    return {
        "rouge1": rouge_result["rouge1"],
        "rouge2": rouge_result["rouge2"],
        "rougeL": rouge_result["rougeL"],
        "bleu": bleu_result["score"],
        "inference_ms": sum(times) / len(times)
    }
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from transformers import TrainingArguments, Trainer
from peft import prepare_model_for_kbit_training, LoraConfig, get_peft_model

def clean_memory():
    gc.collect()
    torch.cuda.empty_cache()

def run_tinyllama_3_seed():
    model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
    short_name = "TinyLlama-1.1B"
    seeds = [42, 123, 2026]

    all_results = []

    for seed in seeds:
        print("="*80)
        print(f"MODEL: {short_name} | SEED: {seed}")
        print("="*80)

        set_seed(seed)
        random.seed(seed)
        np.random.seed(seed)
        torch.manual_seed(seed)

        clean_memory()
        torch.cuda.reset_peak_memory_stats()

        tokenizer = AutoTokenizer.from_pretrained(model_name)

        if tokenizer.pad_token is None:
            tokenizer.pad_token = tokenizer.eos_token

        bnb_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_compute_dtype=torch.float16,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_use_double_quant=True
        )

        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            quantization_config=bnb_config,
            device_map="auto"
        )

        model = prepare_model_for_kbit_training(model)

        train_dataset = TextDataset(train_texts, tokenizer, max_length=1024)
        val_dataset = TextDataset(val_texts, tokenizer, max_length=1024)

        lora_config = LoraConfig(
            r=8,
            lora_alpha=16,
            target_modules=["q_proj", "v_proj"],
            lora_dropout=0.05,
            bias="none",
            task_type="CAUSAL_LM"
        )

        model = get_peft_model(model, lora_config)

        output_dir = f"/content/drive/MyDrive/llm2/seed_runs/{short_name}_seed_{seed}"

        training_args = TrainingArguments(
            output_dir=output_dir,
            num_train_epochs=3,
            per_device_train_batch_size=1,
            per_device_eval_batch_size=1,
            gradient_accumulation_steps=4,
            learning_rate=2e-4,
            logging_steps=1,
            eval_strategy="epoch",
            save_strategy="no",
            fp16=True,
            report_to="none",
            seed=seed
        )

        trainer = Trainer(
            model=model,
            args=training_args,
            train_dataset=train_dataset,
            eval_dataset=val_dataset
        )

        start_train = time.time()
        train_output = trainer.train()
        end_train = time.time()

        train_time = end_train - start_train
        train_loss = train_output.metrics["train_loss"]

        eval_result = trainer.evaluate()
        val_loss = eval_result["eval_loss"]

        metrics = evaluate_model(model, tokenizer)

        max_memory_gb = torch.cuda.max_memory_allocated() / (1024**3)

        row = {
            "model": short_name,
            "seed": seed,
            "train_loss": train_loss,
            "val_loss": val_loss,
            "train_time_sec": train_time,
            "gpu_memory_gb": max_memory_gb,
            **metrics
        }

        all_results.append(row)
        print(row)

        del model
        del trainer
        del tokenizer
        clean_memory()

    result_df = pd.DataFrame(all_results)

    save_path = f"/content/drive/MyDrive/llm2/seed_runs/{short_name}_3seed_results.csv"
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    result_df.to_csv(save_path, index=False, encoding="utf-8-sig")

    print("Kaydedildi:", save_path)
    display(result_df)

    summary_df = result_df.drop(columns=["seed"]).groupby("model").agg(["mean", "std"])
    display(summary_df)

    return result_df, summary_df
```

```python
tiny_3seed, tiny_3seed_summary = run_tinyllama_3_seed()
```
