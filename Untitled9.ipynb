```python
import os
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from peft import PeftModel

BASE_DIR = "/content/drive/MyDrive/llm2"
DATASET_DIR = f"{BASE_DIR}/datasets"
RESULTS_DIR = f"{BASE_DIR}/results"

os.makedirs(RESULTS_DIR, exist_ok=True)

TEST_PATH = f"{DATASET_DIR}/test_40.csv"

BASE_MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"
ADAPTER_PATH = f"{BASE_DIR}/models/qwen25_15b_danistay_40"

print("Test var mı:", os.path.exists(TEST_PATH))
print("Adapter var mı:", os.path.exists(ADAPTER_PATH))
```

```python
test_df = pd.read_csv(TEST_PATH)

print(test_df.shape)
display(test_df[["id", "category"]])
```

```python
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True
)

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto"
)

model = PeftModel.from_pretrained(
    base_model,
    ADAPTER_PATH
)

model.eval()
print("Qwen fine-tuned model yüklendi.")
```

```python
def make_prompt_only(row):
    return f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}

Özet:
"""

def generate_summary(text, max_new_tokens=250):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.2,
            no_repeat_ngram_size=4,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = outputs[0][inputs["input_ids"].shape[-1]:]
    return tokenizer.decode(generated, skip_special_tokens=True).strip()
```

```python
rows = []

for _, row in test_df.iterrows():
    prompt = make_prompt_only(row)
    pred = generate_summary(prompt)

    rows.append({
        "id": row["id"],
        "category": row["category"],
        "reference_summary": row["summary"],
        "qwen_prediction": pred
    })

error_df = pd.DataFrame(rows)

display(error_df)
```

```python
save_path = f"{RESULTS_DIR}/qwen_error_analysis_predictions.csv"

error_df.to_csv(
    save_path,
    index=False,
    encoding="utf-8-sig"
)

print("Kaydedildi:", save_path)
```

```python
def make_qwen_chat_prompt(row):
    messages = [
        {
            "role": "system",
            "content": "Sen Türkçe hukuki karar özetleme konusunda uzman bir asistansın."
        },
        {
            "role": "user",
            "content": f"""Aşağıdaki Danıştay kararını 6-8 cümlelik Türkçe hukuki özet haline getir.

Karar:
{row["model_input"]}"""
        }
    ]

    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

row = test_df.iloc[0]
prompt = make_qwen_chat_prompt(row)

pred = generate_summary(prompt)

print(pred)
```
